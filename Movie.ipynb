```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('/content/movies_metadata.csv')
df.head()
```

```python
df.info()
```

```python
df=df[['title','genres','overview','tagline','vote_average','vote_count']]
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['title'], inplace=True)
```

```python
df['overview']=df['overview'].fillna(' ')
df['tagline']=df['tagline'].fillna(' ')
```

```python
df['vote_average']=df['vote_average'].fillna(df['vote_average'].mean())
df['vote_count']=df['vote_count'].fillna(df['vote_count'].mean())
```

```python
df.isnull().sum()
```

```python
import ast

def extract_names(text):
    names = []
    for item in ast.literal_eval(text):
        names.append(item['name'])
    return names

df['genres'] = df['genres'].apply(extract_names)
```

```python
df['text'] = df['genres'].apply(lambda X: ' '.join(X)) + ' ' + df['overview'] + " " + df['tagline']
```

```python
df.head()
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
```

```python
def preproces_text(text):
  test=str(text).lower()
  test=re.sub(r"[^\w\s]", "", text)
  words=text.split()
  words=[word for word in words if word is not stop_words ]
  words=[lemmatizer.lemmatize(word)for word in words]
  return " ".join(words)

df['text']=df['text'].apply(preproces_text)
```

```python
df.head()
```

```python
df=df.reset_index(drop=True)
```

```python
indices= pd.Series(df.index,index=df['title']).drop_duplicates()
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_features=50000,stop_words='english')
matrix =tfidf.fit_transform(df['text'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommed(title,n=10):
 if title not in indices:
  return['Movie not found']

 idx=indices[title]
 sim_score=cosine_similarity(matrix[idx],matrix).flatten()
 score=sim_score.argsort()[::-1][1:n+1]
 return df['title'].iloc[score]
```

```python
recommed('Jumanji',5)
```

## Save Preprocessed Data and TF-IDF Components

To ensure that the Streamlit application can load the necessary components efficiently, we'll save the processed DataFrame, the TF-IDF matrix, and the movie title-to-index mapping to pickle files.

```python
import pickle
import joblib

# Save the preprocessed DataFrame
with open('df.pkl', 'wb') as f:
    pickle.dump(df, f)
```

```python
# Save the TF-IDF matrix
with open('tfidf_matrix.pkl', 'wb') as f:
    pickle.dump(matrix, f)
```

```python
# Save the movie title to index mapping
with open('indices.pkl', 'wb') as f:
    pickle.dump(indices, f)

print("df.pkl, tfidf_matrix.pkl, and indices.pkl saved successfully.")
```
